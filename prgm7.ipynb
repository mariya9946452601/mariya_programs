```python
color_set1 = {"red", "green", "blue"}
color_set2 = {"yellow", "blue", "black"}

print("Colors from both sets:")
print(color_set1 | color_set2)
```

```python
color_set1 = {"red", "green", "blue", "yellow"}
color_set2 = {"blue", "yellow", "black"}

print("Colors in Set 1 but not in Set 2:")
print(color_set1 - color_set2)
```
