```python
numbers = input("Enter comma-separated numbers: ")

num_tuple = tuple(float(x.strip()) for x in numbers.split(","))

average = sum(num_tuple) / len(num_tuple)

print("Tuple:", num_tuple)
print("Average:", average)
```
