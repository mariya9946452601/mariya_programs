```python
#Exception Handling using ValueError
try:
    num = int(input("Enter a positive number: "))
 
    if num <= 0:
        raise ValueError("Invalid input! Enter a positive number.")
        
    print("You entered:", num)
    
except ValueError as e:
    print("Error:", e)
```
