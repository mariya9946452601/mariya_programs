```python
# Create a tuple
t = tuple(input("Enter tuple elements separated by spaces: ").split())

# Display the 3rd element and 3rd element from last
print("3rd element:", t[2])
print("3rd element from last:", t[-3])
```

```python
# Accept comma-separated numbers from the user
numbers = input("Enter comma-separated numbers: ")

# Convert the input into a tuple of numbers
t = tuple(map(float, numbers.split(',')))

# Calculate the average
avg = sum(t) / len(t)

# Display the tuple and average
print("Tuple:", t)
print("Average:", avg)
```
