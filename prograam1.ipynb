```python
# compute n+nn+nnn

num = int(input("Enter a number: "))

num1 = num
num2 = num * 10 + num1
num3 = num * 100 + num2

print(num1,"+",num2,"+",num3,"=",num1+num2+num3)
```

```python
# area of circle by inputting radius

import math

r = int(input("Enter the radius of the circle: "))
area = math.pi * r * r

print("Area of the circle:"f"{area:.2f}")
```

```python
# biggest of three input number using max()

n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))
n3 = int(input("Enter third number: "))

max = max(n1, n2, n3)

print("Maximum number:", max)
```
