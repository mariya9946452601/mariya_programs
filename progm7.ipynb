```python
# Input color sets
colorset1 = set(input("Enter colors in color set 1: ").split())
colorset2 = set(input("Enter colors in color set 2: ").split())

# Print colors in set 1 but not in set 2
print("Colors in color set 1 but not in color set 2:")
print(colorset1 - colorset2)
```

```python
# Input color sets
colorset1 = set(input("Enter colors in color set 1: ").split())
colorset2 = set(input("Enter colors in color set 2: ").split())

# Print all colors from both sets
print("All colors from color set 1 and color set 2:")
print(colorset1 | colorset2)
```
