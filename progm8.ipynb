```python
# Accept names and roll numbers of 5 students
students = {}

for i in range(5):
    name = input("Enter student name: ")
    roll = int(input("Enter roll number: "))
    students[roll] = name

# Sort in ascending order of roll number
ascending = dict(sorted(students.items()))

# Sort in descending order of roll number
descending = dict(sorted(students.items(), reverse=True))

print("\nDictionary in ascending order:")
print(ascending)

print("\nDictionary in descending order:")
print(descending)
```

```python
# Accept the first dictionary (3 students)
dic1 = {}

for i in range(3):
    name = input("Enter student name: ")
    roll = int(input("Enter roll number: "))
    dic1[roll] = name

# Accept the second dictionary (2 students)
dic2 = {}

for i in range(2):
    name = input("Enter student name: ")
    roll = int(input("Enter roll number: "))
    dic2[roll] = name

# Merge the dictionaries using update()
dic1.update(dic2)

# Display the merged dictionary
print("Merged dictionary:", dic1)
```
